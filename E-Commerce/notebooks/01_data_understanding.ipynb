```python
missing_values=df.isna().sum()
print(missing_values)
```

```python
missing_table = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_table.sort_values(
    by="Missing Count",
    ascending=False
)
```
