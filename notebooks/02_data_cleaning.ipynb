```python
import pandas as pd
import numpy as np

file_path = "../data/raw/E Commerce Dataset.xlsx"

df = pd.read_excel(file_path, sheet_name="E Comm")

print(df.shape)
```

```python
df_clean = df.copy()
```

```python
print("duplicated rows:",df_clean.duplicated().sum())
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
print("after removing the duplicated rows:",df_clean.shape)
```

```python
missing = pd.DataFrame({
    "Missing Count": df_clean.isnull().sum(),
    "Missing %": df_clean.isnull().mean() * 100
})

missing = missing.sort_values(
    "Missing Count",
    ascending=False
)

missing
```

```python
numerical_cols = df_clean.select_dtypes(
    include=np.number
).columns
```

```python
for col in numerical_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
num_cols_to_impute = [
    col for col in numerical_cols
    if col not in ["CustomerID", "Churn"]
]

for col in num_cols_to_impute:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
categorical_cols = df_clean.select_dtypes(
    include="object"
).columns.tolist()

print(categorical_cols)
```

```python
df_clean[categorical_cols].isnull().sum()
```

```python
for col in categorical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[
        (df[column] < lower) |
        (df[column] > upper)
    ]
```

```python
for col in [
    "Tenure",
    "WarehouseToHome",
    "HourSpendOnApp",
    "OrderAmountHikeFromlastYear",
    "CouponUsed",
    "OrderCount",
    "DaySinceLastOrder",
    "CashbackAmount"
]:
    print(col, len(find_outliers(df_clean, col)))
```
