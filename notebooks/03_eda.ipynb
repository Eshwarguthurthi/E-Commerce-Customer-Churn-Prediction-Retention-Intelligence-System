```python
def churn_rate_by_column(df, column):
    result = (
        df.groupby(column)["Churn"]
        .agg(["count", "mean"])
        .reset_index()
    )

    result["ChurnRate_%"] = result["mean"] * 100
    result = result.drop(columns="mean")

    return result.sort_values(
        "ChurnRate_%",
        ascending=False
    )
    
```

```python
churn_rate_by_column(df, "PreferredLoginDevice")
```
